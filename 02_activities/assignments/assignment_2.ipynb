# Explanation
In this example, the binary tree is constructed with the root node having a value of 4. The left child of the root is 2, and the right child is 6. The left subtree contains a duplicate value of 2, and the right subtree has a duplicate value of 6. During the level-order traversal, the first duplicate encountered is 2, so the function should return 2.


-   Copy the solution your partner wrote. 


````python
# Your answer here
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def is_duplicate(root: TreeNode) -> int:
    if not root:
        return -1
    seen = set()
    queue = [root]
    i = 0

    while i < len(queue):
        node = queue[i]
        i += 1

        if node.val in seen:
            return node.val
        
        seen.add(node.val)

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)

    return -1

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(3)
root.right.right = TreeNode(2)

print(is_duplicate(root))
````

We're given the root of a binary tree, and the objective is to check if there's any duplicate value. If we find duplicates, return the one that appears closest to the root (i.e., the one encountered first in a level-order traversal). If no duplicates are present, return -1.

# Explanation for Algorithm
We manually traverse through each node and keep track of the values we have seen before. It will break ties between multiple duplicates based on depth (which one is closest to the node), which is inherently determined by the first set of duplicates found.

# Explanation of Time and Space Complexity
All steps outside of, and within, the `while` loop have an O(1) efficiency, and the `while` loop itself has O(n) time efficiency; hence, the algorithm has an O(n) time efficiency. The `queue` list and `seen` set's size are each bounded by the number of nodes in the tree, so the algorithm has O(n) space efficiency.

# Alternative explanation
Instead of breadth-first (level-order) traversal, you can instead use DFS:

Traverse the tree by diving deep into one branch at a time, keeping track of the depth of the first of each unique value.

1. Keep track of the depth of each node value found in a dict (hash map)
2. Keep track of the duplicate value that is closest to the root node (bestDuplicate)
3. Recursively visit each node in the binary tree in a depth-first order.
4. At each node:
    1. Add the value and depth to the dict if the current node value hasn't been seen before
    2. If current node value is a duplicate, compare the current node's depth to the best duplicate's depth. If the current depth is less than that of the current bestDuplicate, update bestDuplicate to the current node’s value.
