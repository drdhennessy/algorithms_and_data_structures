
-   Code the solution to your assigned problem in Python (code chunk). Try to find the best time and space complexity solution!


```python
# Your answer here
from typing import List

def missing_num(nums: List[int]) -> List[int]:
    if not nums:
        return -1  # Edge case: empty list
    
    n = max(nums)  # Find the maximum number in the list
    full_range = set(range(n + 1))  # Generate the full range of numbers
    present_numbers = set(nums)  # Convert list to set to remove duplicates
    
    missing = sorted(full_range - present_numbers)  # Find missing numbers and sort them
    
    return missing if missing else -1  # Return missing numbers or -1 if none are missing

# Testing provided examples
print(missing_num([0, 2]))                # Expected output: [1]
print(missing_num([5, 0, 1]))             # Expected output: [2, 3, 4]
print(missing_num([6, 8, 2, 3, 5, 7, 0, 1, 10]))  # Expected output: [4, 9]

# Testing new examples
print(missing_num([3, 0, 1, 3, 5, 5]))  # Expected output: [2, 4]
print(missing_num([4, 2, 0, 3, 1]))     # Expected output: -1
```


-   Explain why your solution works


# Your answer here:
1. Identifies the Maximum Value (n)
    This allows us to determine the full range [0, n] that should be present.
    
2. Uses a Set to Handle Duplicates Efficiently
    Converting the list to a set removes any duplicate values, ensuring that each    number is considered only once.

3. Finds Missing Numbers Using Set Difference
    full_range - present_numbers quickly identifies the numbers that are missing.

4. Sorts the Missing Numbers
    Ensures the output is in ascending order for clarity.

5. Handles Edge Cases
    If no numbers are missing, returns -1 instead of an empty list.
    If the input list is empty, returns -1 immediately.


-   Explain the problem’s time and space complexity


# Your answer here
Time & Space Complexity

Time Complexity: 
    Finding the maximum number takes O(n).
    Creating the full range takes O(n).
    Converting the list to a set takes O(n).
    Finding missing numbers with set subtraction takes O(n).
    Sorting takes O(k log k), where k is the number of missing elements.
    The worst case would be O(n + k log k).

Space Complexity: O(n)
    It is O(n) for storing the full range and given numbers in sets.
    It is O(k) for storing missing numbers. 
    Overall, it would be O(n) in practice.
